```python

delivery_time_hours_test = (
    pd.to_datetime(df["delivery_time_hours"])
    - pd.Timestamp("1970-01-01")
).dt.total_seconds() / 3600

delivery_time_hours_test.describe()
```

```python
delivery_time_recovered = delivery_time_test.dt.nanosecond
```
